```python
dados = []
with open('ocorrencias_toledo.csv', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    dados = [linha for linha in reader]
print(f'Total de registros: {len(dados)}')
print('Exemplo:', dados[0])
```

### Total de ocorrências

```python
print(len(dados))
```

### Ocorrências no bairro Centro

```python
print(len([oc for oc in dados if oc['bairro'].lower() == 'centro']))
```

### Ocorrências no bairro Vila Nova

```python
print(len([oc for oc in dados if oc['bairro'].lower() == 'vila nova']))
```

### Ocorrências com severidade alta

```python
print(len([oc for oc in dados if oc['severidade'].lower() == 'alta']))
```

### Ocorrências com severidade baixa

```python
print(len([oc for oc in dados if oc['severidade'].lower() == 'baixa']))
```

### Ocorrências com severidade média

```python
print(len([oc for oc in dados if oc['severidade'].lower() == 'média']))
```

### Ocorrências do tipo furto

```python
print(len([oc for oc in dados if oc['tipo'].lower() == 'furto']))
```

### Ocorrências do tipo roubo

```python
print(len([oc for oc in dados if oc['tipo'].lower() == 'roubo']))
```

### Ocorrências do tipo acidente

```python
print(len([oc for oc in dados if oc['tipo'].lower() == 'acidente']))
```

### Ocorrências no Jardim Porto Alegre

```python
print(len([oc for oc in dados if oc['bairro'].lower() == 'jardim porto alegre']))
```

### Ocorrências com descrição maior que 50 caracteres

```python
print(len([oc for oc in dados if 'descricao' in oc and len(oc['descricao']) > 50]))
```

### Ocorrências em junho

```python
print(len([oc for oc in dados if '-06-' in oc['data']]))
```

### Ocorrências em dias pares

```python
print(len([oc for oc in dados if int(oc['data'].split('-')[2]) % 2 == 0]))
```

### Ocorrências em dias ímpares

```python
print(len([oc for oc in dados if int(oc['data'].split('-')[2]) % 2 != 0]))
```

### Ocorrências com bairro iniciando com 'vila'

```python
print(len([oc for oc in dados if oc['bairro'].lower().startswith('vila')]))
```

### Ocorrências com tipo iniciando com 'f'

```python
print(len([oc for oc in dados if oc['tipo'].lower().startswith('f')]))
```

### Ocorrências com severidade diferente de baixa

```python
print(len([oc for oc in dados if oc['severidade'].lower() != 'baixa']))
```

### Tipos únicos de ocorrência

```python
print(set([oc['tipo'] for oc in dados]))
```

### Bairros únicos

```python
print(set([oc['bairro'] for oc in dados]))
```

### Quantidade por tipo

```python
print(Counter([oc['tipo'] for oc in dados]))
```

### Quantidade por bairro

```python
print(Counter([oc['bairro'] for oc in dados]))
```

### Quantidade por severidade

```python
print(Counter([oc['severidade'] for oc in dados]))
```

### Tipo mais comum

```python
print(Counter([oc['tipo'] for oc in dados]).most_common(1))
```

### Tipo menos comum

```python
print(Counter([oc['tipo'] for oc in dados]).most_common()[-1])
```

### Severidade mais comum

```python
print(Counter([oc['severidade'] for oc in dados]).most_common(1))
```

### Severidade menos comum

```python
print(Counter([oc['severidade'] for oc in dados]).most_common()[-1])
```

### Ocorrências com descrição contendo 'acidente'

```python
print(len([
    oc for oc in dados
    if 'acidente' in oc.get('descricao', '').lower()
]))
```

### Ocorrências no bairro Jardim Europa

### Ocorrências do tipo violência doméstica

```python
print(len([oc for oc in dados if oc['tipo'].lower() == 'violência doméstica']))
```

### Ocorrências em ordem cronológica

```python
print(sorted(dados, key=lambda oc: oc['data'])[:3])
```
